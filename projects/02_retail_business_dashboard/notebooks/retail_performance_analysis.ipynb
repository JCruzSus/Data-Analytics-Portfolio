```python
from pathlib import Path
import pandas as pd, subprocess, sys
ROOT = Path.cwd()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
```

```python
subprocess.run([sys.executable, str(ROOT/'src'/'analyze_and_visualize.py')], check=True)
df = pd.read_csv(ROOT/'data'/'retail_transactions.csv', parse_dates=['order_date'])
```

```python
display(pd.read_csv(ROOT/'outputs'/'kpis.csv'))
display(pd.read_csv(ROOT/'outputs'/'regional_performance.csv').sort_values('profit', ascending=False))
display(pd.read_csv(ROOT/'outputs'/'category_performance.csv').sort_values('profit', ascending=False))
```

```python
import plotly.express as px
monthly = df.assign(month=df.order_date.dt.to_period('M').astype(str)).groupby('month', as_index=False).sales.sum()
px.line(monthly, x='month', y='sales', title='Monthly Revenue').show()
```
