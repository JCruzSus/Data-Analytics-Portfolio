# Retail Business Performance

Generate the seeded synthetic transactions, inspect revenue and profit KPIs, and review product/region results.

```python
from pathlib import Path
import pandas as pd, subprocess, sys
ROOT = Path.cwd()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
```

```python
subprocess.run([sys.executable, str(ROOT/'src'/'generate_data.py')], check=True)
df = pd.read_csv(ROOT/'data'/'retail_transactions.csv', parse_dates=['order_date'])
```

```python
print('Revenue:', df.sales.sum())
print('Gross profit:', df.profit.sum())
print('Gross margin:', df.profit.sum()/df.sales.sum())
display(df.groupby('region').agg(revenue=('sales','sum'), profit=('profit','sum')).sort_values('profit', ascending=False))
display(df.groupby('category').agg(revenue=('sales','sum'), profit=('profit','sum')).sort_values('profit', ascending=False))
```

```python
import plotly.express as px
monthly = df.assign(month=df.order_date.dt.to_period('M').astype(str)).groupby('month', as_index=False).sales.sum()
px.line(monthly, x='month', y='sales', title='Monthly Revenue').show()
```
