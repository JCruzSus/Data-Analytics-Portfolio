# Employee Attrition Analysis & Prediction

This notebook uses generated synthetic employee records, not IBM's dataset or real HR records. It compares two classifiers and demonstrates imbalanced-class evaluation.

```python
from pathlib import Path
import pandas as pd, subprocess, sys
ROOT = Path.cwd()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
```

```python
subprocess.run([sys.executable, str(ROOT/'src'/'generate_synthetic_data.py')], check=True)
df = pd.read_csv(ROOT/'data'/'synthetic_hr_attrition.csv')
```

```python
df['AttritionFlag'] = (df.Attrition == 'Yes').astype(int)
print('Rows:', len(df))
print('Attrition rate:', df.AttritionFlag.mean())
display(df.groupby('OverTime').AttritionFlag.agg(['count','mean']))
```
